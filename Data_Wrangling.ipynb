```python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Name": ["Mary", "John", "Amina"]
})

scores = pd.DataFrame({
    "Student_ID": [101, 103, 104],
    "Score": [80, 90, 75]
})

print(students)
print(scores)
```

```python
pd.merge(students, scores, on="Student_ID", how="right")
```

```python
pd.merge(students,scores,on='Student_ID',how='inner')
```

```python
pd.merge(students,scores,on='Student_ID',how='left')
```

```python
pd.merge(students,scores,on='Student_ID',how='outer')
```

```python


# DataFrame 1: Customers
data_customers = {
    'CustomerID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'City': ['New York', 'London', 'Paris', 'New York', 'Tokyo']
}
df_customers = pd.DataFrame(data_customers)

# DataFrame 2: Orders
# Note:
# - CustomerID 101 and 102 have orders (will match inner)
# - CustomerID 106 exists in orders but not customers (will be seen in right/outer)
# - CustomerID 103, 104, 105 exist in customers but have no orders (will be seen in left/outer)
data_orders = {
    'OrderID': ['ORD001', 'ORD002', 'ORD003', 'ORD004', 'ORD005'],
    'CustomerID': [101, 102, 101, 106, 102],
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam'],
    'Quantity': [1, 2, 1, 1, 1]
}
df_orders = pd.DataFrame(data_orders)

print("--- df_customers ---")
print(df_customers)
print("\n--- df_orders ---")
print(df_orders)
```

```python
df_customers.merge(df_orders, on='CustomerID', how='inner')
```

```python
pd.merge(df_customers,df_orders,on='CustomerID',how='right')
```

```python
pd.merge(df_customers,df_orders,on='CustomerID',how='left')
```

```python
pd.merge(df_customers,df_orders,on='CustomerID',how='outer')
```

```python
pd.merge(df_orders,df_customers,on='CustomerID',how='outer')
```
